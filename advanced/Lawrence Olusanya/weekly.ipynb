```python
# Import libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
import pathlib
import tensorflow as tf
from sklearn.preprocessing import MinMaxScaler
```

### Load dataset & pre-processing

```python

dataset_path = "data/Tetuan City power consumption.csv"
data = pd.read_csv(dataset_path, parse_dates=["DateTime"], index_col=["DateTime"])
data.head()
```

```python
data_hr = data.copy()
```

```python
# Drop column
data.drop('diffuse flows', axis = 1, inplace=True)
```

```python
#Resample data
daily = data.resample('D')
weekly = data.resample('W')
hourly = data.resample('h')
monthly = data.resample('ME')
```

### Weekly Prediction

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

def create_sliding_window2(groups, input_days=2, output_days=1, target_columns = ['Zone 1 Power Consumption', 'Zone 2  Power Consumption','Zone 3  Power Consumption']):
    X, y = [], []
    #get indices of target columns
    target_indices = [data.columns.get_loc(col) for col in target_columns]
    for group in groups:
        for i in range(len(group) - input_days - output_days + 1):
            X.append(group.iloc[i:i + input_days].values)
            y.append(group.iloc[i + input_days:i + input_days + output_days, target_indices].values)
    return np.array(X), np.array(y)


#prepare test_group_scaled for model evaluation 
def create_sliding_window2_test(groups, input_days=14, output_days=7, target_columns = ['Zone 1 Power Consumption', 'Zone 2  Power Consumption','Zone 3  Power Consumption']):
    X, y = [], []
    #get indices of target columns
    target_indices = [data.columns.get_loc(col) for col in target_columns]
    for group in groups:
        for i in range(len(group) - input_days - output_days + 1):
            X.append(group.iloc[i:i + input_days].values)
            y.append(group.iloc[i + input_days:i + input_days + output_days, target_indices].values)
    return np.array(X), np.array(y)

def evaluate_model(model, X_test, y_test):
    
    # Predict
    y_pred = model.predict(X_test)
    # Inverse transform predictions (reshape as needed)
    y_pred_flat = y_pred.reshape(-1, 3)
    y_pred_inv = scaler.inverse_transform(np.concatenate([np.zeros((y_pred_flat.shape[0], X_test.shape[2]-3)), y_pred_flat], axis=1))[:, -3:]  # Only get the target columns
    
    
    mae = mean_absolute_error(y_test.flatten(), y_pred.flatten())
    rmse = np.sqrt(mean_squared_error(y_test.flatten(), y_pred.flatten()))
    r2 = r2_score(y_test.flatten(), y_pred.flatten())
  

    print(f'Mean Absolute Error: {mae}')
    print(f'Root Mean Squared Error: {rmse}')
    print(f'R² Score: {r2}')
    
    return y_pred, y_test

all_input_days= [2,3,4,5,6,7]

weekly_groups = [group for _, group in data.resample('W-MON')]

#split weekly_group into training and test set 0.8,0.2
train_size = int(len(weekly_groups) * 0.8)
train_groups = weekly_groups[:train_size]

for input_days in all_input_days:




    #normalise the train_groups using minmaxscaler
    scaler = MinMaxScaler()
    train_groups_scaled = [pd.DataFrame(scaler.fit_transform(group), index=group.index, columns=group.columns) for group in train_groups]  
 
    # pastrain_group_scaled to lstm encoder
    X_train, y_train = create_sliding_window2(train_groups_scaled, input_days=input_days, output_days=1, target_columns=['Zone 1 Power Consumption', 'Zone 2  Power Consumption','Zone 3  Power Consumption'])

    # Define the LSTM model
    from keras.models import Sequential
    from keras.layers import LSTM, Dense, Dropout
    model = Sequential()
    model.add(LSTM(64,  input_shape=(X_train.shape[1], X_train.shape[2]), return_sequences=True, activation='relu'))
    model.add(Dropout(0.2))
    model.add(LSTM(32, return_sequences=False, activation='relu'))
    model.add(Dropout(0.2))

    model.add(Dense(y_train.shape[1] * y_train.shape[2], activation='linear'))  
    model.add(tf.keras.layers.Reshape((y_train.shape[1], y_train.shape[2])))
    model.compile(optimizer='adam', loss='mse')

    # Train the model
    history = model.fit(X_train, y_train, epochs=50, batch_size=32, validation_split=0.2)

    #create test group
    test_groups = weekly_groups[train_size:]

    #normalise the test_groups using minmaxscaler
    test_groups_scaled = [pd.DataFrame(scaler.transform(group), index=group.index, columns=group.columns) for group in test_groups]

    X_test, y_test = create_sliding_window2_test(test_groups_scaled, input_days=input_days, output_days=1, target_columns=['Zone 1 Power Consumption', 'Zone 2  Power Consumption','Zone 3  Power Consumption'])

    #evalauete prediction against the actual using rrmse or mae
    from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


    # Evaluate and plot predictions for all zones with correct shape handling
    y_pred, y_test = evaluate_model(model, X_test, y_test)
    
   
    n_features = X_test.shape[2]  
    n_targets = 3 
    
    # Flatten predictions and targets to 2D (samples*timesteps, n_targets)
    y_pred_flat = y_pred.reshape(-1, n_targets)
    y_test_flat = y_test.reshape(-1, n_targets)
    
    # Pad with zeros for non-target columns to match scaler input shape
    y_pred_padded = np.concatenate([np.zeros((y_pred_flat.shape[0], n_features - n_targets)), y_pred_flat], axis=1)
    y_test_padded = np.concatenate([np.zeros((y_test_flat.shape[0], n_features - n_targets)), y_test_flat], axis=1)
    
    # Inverse transform
    try:
        y_pred_inv = scaler.inverse_transform(y_pred_padded)[:, -n_targets:]
        y_test_inv = scaler.inverse_transform(y_test_padded)[:, -n_targets:]
    except Exception as e:
        print('Inverse transform error:', e)
        y_pred_inv = y_pred_flat
        y_test_inv = y_test_flat
        
    # Calculate metrics for each target (zone)
   
    for i, zone in enumerate(['Zone 1', 'Zone 2', 'Zone 3']):
        rmse = np.sqrt(mean_squared_error(y_test_inv[:, i], y_pred_inv[:, i]))
        mae = mean_absolute_error(y_test_inv[:, i], y_pred_inv[:, i])
        r2 = r2_score(y_test_inv[:, i], y_pred_inv[:, i])
        print(f'{zone} - RMSE: {rmse:.2f}, MAE: {mae:.2f}, R²: {r2:.2f}')
        # Plot actual vs predicted
        plt.figure(figsize=(12, 4))
        plt.plot(y_test_inv[:, i], label='Actual')
        plt.plot(y_pred_inv[:, i], label='Predicted')
        plt.title(f'{zone} Power Consumption: Actual vs Predicted')
        plt.xlabel('Time step')
        plt.ylabel('Power Consumption')
        plt.legend()
        plt.show()
    
    plt.plot(history.epoch, history.history['loss'], label='loss')
    plt.plot(history.epoch, history.history['val_loss'], label = 'val loss')
    plt.xlabel('Epochs')
    plt.ylabel('losses')
    plt.title('Plot Losses versus Epoch')
    plt.legend()
    plt.show()


                    
```
