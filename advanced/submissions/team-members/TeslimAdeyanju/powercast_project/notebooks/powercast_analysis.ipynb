The PowerCast Advanced Track project focuses on forecasting electricity consumption in Tetouan City using deep learning techniques applied to time-series environmental data. ## 🔍 1.2 Project Objective
___



The dataset used in this project is the [UCI Tetouan Power Consumption Dataset](https://archive.ics.uci.edu/dataset/849/power+consumption+of+tetouan+city). It contains time-series records of electricity usage across three urban zones in Tetouan City, along with environmental features such as:

- **Timestamp**: Date and time of each record
- **Temperature**: Ambient temperature readings
- **Humidity**: Relative humidity levels
- **Wind Speed**: Measured wind speed
- **Solar Radiation**: Solar energy received

The primary objective is to leverage these features to forecast future power consumption, uncovering temporal trends, seasonality, and relationships between environmental conditions and energy demand.

````python
# Week 2: Feature Engineering & Deep Learning Preparation - Using Modules
# ========================================================================

print_section_header("WEEK 2: FEATURE ENGINEERING & PREPARATION", level=1)

# Prepare numeric data for processing
numeric_cols = ['power_consumption', 'temperature', 'humidity', 'wind_speed', 'hour', 'day_of_week', 'month']
numeric_data = data[numeric_cols].values

print_info(f"Prepared numeric data shape: {numeric_data.shape}")

# 1. Analyze Optimal Lookback Windows
print_section_header("LOOKBACK WINDOW ANALYSIS", level=2)
windows_to_test = [24, 48, 72, 96, 144, 168]
print_info(f"Testing lookback windows: {windows_to_test}")

window_results = analyze_lookback_windows(
    data=numeric_data, 
    target_column=-1,  # Using power_consumption as target
    windows=windows_to_test
)

# Find optimal window
best_window = max(window_results.keys(), key=lambda w: window_results[w]['efficiency_score'])
print_success(f"Optimal lookback window: {best_window} hours")

for window, metrics in window_results.items():
    print(f"  Window {window:3d}: {metrics['samples_generated']:4d} samples, "
          f"efficiency: {metrics['efficiency_score']:.3f}")

# 2. Setup Preprocessing Pipeline
print_section_header("PREPROCESSING PIPELINE SETUP", level=2)

# Initialize preprocessing pipeline with optimal settings
pipeline = PreprocessingPipeline(
    lookback_window=best_window,
    forecast_horizon=1,
    scaling_method='standard'
)

print_info(f"Pipeline configured:")
print(f"  • Lookback window: {best_window} hours")
print(f"  • Forecast horizon: 1 step")
print(f"  • Scaling method: standard")

# 3. Execute Preprocessing Pipeline
print_section_header("PREPROCESSING EXECUTION", level=2)

feature_names = ['power_consumption', 'temperature', 'humidity', 'wind_speed', 'hour', 'day_of_week', 'month']

processed_data = pipeline.fit_transform(
    data=numeric_data,
    target_column=0,  # power_consumption is first column
    feature_names=feature_names
)

# Extract processed datasets
X_train = processed_data['X_train']
X_val = processed_data['X_val']
X_test = processed_data['X_test']
y_train = processed_data['y_train']
y_val = processed_data['y_val']
y_test = processed_data['y_test']

print_success("Preprocessing pipeline completed successfully!")
print(f"  • Training set: {X_train.shape}")
print(f"  • Validation set: {X_val.shape}")  
print(f"  • Test set: {X_test.shape}")

# 4. Pipeline Summary
print_section_header("PREPROCESSING SUMMARY", level=2)
summary = pipeline.get_pipeline_summary()
print(summary)

print_success("Week 2 Feature Engineering completed using organized modules!")
````

## 3.1 Sequence Construction & Lookback Windows

Time series forecasting requires transforming our data into sequences where the model learns from historical patterns to predict future values. We'll explore different lookback window sizes and sequence structures optimized for deep learning architectures.

````python
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings('ignore')

class SequenceConstructor:
    """Class for constructing time series sequences for deep learning models"""
    
    def __init__(self, lookback_window=144, forecast_horizon=1):
        """
        Initialize sequence constructor
        
        Parameters:
        - lookback_window: Number of time steps to look back (default: 144 = 24 hours)
        - forecast_horizon: Number of time steps to forecast (default: 1)
        """
        self.lookback_window = lookback_window
        self.forecast_horizon = forecast_horizon
        
    def create_sequences(self, data, target_column):
        """
        Create sequences from time series data
        
        Parameters:
        - data: DataFrame with time series data
        - target_column: Column name to predict
        
        Returns:
        - X: Input sequences (samples, time_steps, features)
        - y: Target values (samples, forecast_horizon)
        """
        # Ensure data is sorted by time
        data_sorted = data.sort_index()
        
        # Get feature columns (all columns except target)
        feature_columns = [col for col in data_sorted.columns if col != target_column]
        
        X, y = [], []
        
        for i in range(len(data_sorted) - self.lookback_window - self.forecast_horizon + 1):
            # Input sequence (lookback_window time steps)
            X_seq = data_sorted[feature_columns].iloc[i:(i + self.lookback_window)].values
            
            # Target value(s) (forecast_horizon time steps ahead)
            y_seq = data_sorted[target_column].iloc[
                (i + self.lookback_window):(i + self.lookback_window + self.forecast_horizon)
            ].values
            
            X.append(X_seq)
            y.append(y_seq)
        
        return np.array(X), np.array(y)
    
    def analyze_sequence_patterns(self, data, target_column, lookback_windows=[72, 144, 288, 432]):
        """
        Analyze different lookback window sizes and their impact on data availability
        
        Parameters:
        - data: DataFrame with time series data
        - target_column: Column name to predict
        - lookback_windows: List of different lookback window sizes to test
        
        Returns:
        - Dictionary with analysis results
        """
        results = {}
        total_samples = len(data)
        
        print("=== Lookback Window Analysis ===")
        print(f"Total available data points: {total_samples:,}")
        print(f"Data frequency: {data.index.freq or '10 minutes (inferred)'}")
        print()
        
        for window in lookback_windows:
            # Calculate available sequences
            available_sequences = total_samples - window - self.forecast_horizon + 1
            data_usage_percent = (available_sequences / total_samples) * 100
            
            # Time coverage
            hours_coverage = window * 10 / 60  # 10-minute intervals to hours
            days_coverage = hours_coverage / 24
            
            results[window] = {
                'available_sequences': max(0, available_sequences),
                'data_usage_percent': max(0, data_usage_percent),
                'hours_coverage': hours_coverage,
                'days_coverage': days_coverage,
                'window_size': window
            }
            
            print(f"Lookback Window: {window} steps ({hours_coverage:.1f} hours / {days_coverage:.1f} days)")
            print(f"  Available sequences: {max(0, available_sequences):,}")
            print(f"  Data utilization: {max(0, data_usage_percent):.1f}%")
            print()
        
        return results

# Initialize sequence constructor
sequence_constructor = SequenceConstructor()

# Analyze different lookback window sizes
print("PowerCast Dataset - Sequence Construction Analysis")
print("=" * 60)

# Analyze patterns with different lookback windows
window_analysis = sequence_constructor.analyze_sequence_patterns(
    df, 'zone1_power_consumption', 
    lookback_windows=[72, 144, 288, 432, 576]  # 12h, 24h, 48h, 72h, 96h
)
````

## 3.2 Feature Scaling & Transformation

Neural networks perform better with normalized inputs. We'll implement multiple scaling strategies and analyze their impact on different feature types to ensure optimal model convergence and performance.

````python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
from sklearn.preprocessing import PowerTransformer, QuantileTransformer
import scipy.stats as stats

class FeatureScaler:
    """Comprehensive feature scaling and transformation for time series data"""
    
    def __init__(self):
        self.scalers = {}
        self.feature_stats = {}
        
    def analyze_feature_distributions(self, data, features):
        """Analyze feature distributions to determine optimal scaling strategy"""
        analysis_results = {}
        
        fig, axes = plt.subplots(2, 3, figsize=(18, 10))
        axes = axes.flatten()
        
        for i, feature in enumerate(features[:6]):  # Analyze first 6 features
            feature_data = data[feature].dropna()
            
            # Basic statistics
            stats_dict = {
                'mean': feature_data.mean(),
                'std': feature_data.std(),
                'min': feature_data.min(),
                'max': feature_data.max(),
                'skewness': stats.skew(feature_data),
                'kurtosis': stats.kurtosis(feature_data),
                'range': feature_data.max() - feature_data.min()
            }
            
            # Normality test
            _, p_value = stats.normaltest(feature_data.sample(min(5000, len(feature_data))))
            stats_dict['is_normal'] = p_value > 0.05
            
            analysis_results[feature] = stats_dict
            
            # Plot distribution
            axes[i].hist(feature_data, bins=50, alpha=0.7, color='skyblue', edgecolor='black')
            axes[i].set_title(f'{feature}\nSkew: {stats_dict["skewness"]:.2f}, Normal: {stats_dict["is_normal"]}')
            axes[i].set_xlabel('Value')
            axes[i].set_ylabel('Frequency')
            axes[i].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.suptitle('Feature Distribution Analysis for Scaling Strategy', y=1.02, fontsize=16)
        plt.show()
        
        return analysis_results
    
    def fit_multiple_scalers(self, data, features):
        """Fit multiple scaling methods to compare their effectiveness"""
        scaling_methods = {
            'StandardScaler': StandardScaler(),
            'MinMaxScaler': MinMaxScaler(),
            'RobustScaler': RobustScaler(),
            'PowerTransformer': PowerTransformer(method='yeo-johnson'),
            'QuantileTransformer': QuantileTransformer(output_distribution='normal')
        }
        
        fitted_scalers = {}
        scaling_results = {}
        
        for method_name, scaler in scaling_methods.items():
            try:
                # Fit scaler
                fitted_scaler = scaler.fit(data[features])
                fitted_scalers[method_name] = fitted_scaler
                
                # Transform data
                scaled_data = fitted_scaler.transform(data[features])
                scaled_df = pd.DataFrame(scaled_data, columns=features, index=data.index)
                
                # Calculate scaling statistics
                scaling_stats = {}
                for feature in features:
                    feature_scaled = scaled_df[feature]
                    scaling_stats[feature] = {
                        'mean': feature_scaled.mean(),
                        'std': feature_scaled.std(),
                        'min': feature_scaled.min(),
                        'max': feature_scaled.max(),
                        'range': feature_scaled.max() - feature_scaled.min()
                    }
                
                scaling_results[method_name] = {
                    'scaler': fitted_scaler,
                    'stats': scaling_stats,
                    'data': scaled_df
                }
                
            except Exception as e:
                print(f"Warning: {method_name} failed: {e}")
                continue
        
        self.scalers = fitted_scalers
        return scaling_results
    
    def visualize_scaling_comparison(self, original_data, scaling_results, feature_subset=None):
        """Compare different scaling methods visually"""
        if feature_subset is None:
            feature_subset = original_data.columns[:3]  # First 3 features
        
        n_features = len(feature_subset)
        n_methods = len(scaling_results)
        
        fig, axes = plt.subplots(n_features, n_methods + 1, figsize=(20, 4 * n_features))
        if n_features == 1:
            axes = axes.reshape(1, -1)
        
        for i, feature in enumerate(feature_subset):
            # Original data
            axes[i, 0].hist(original_data[feature], bins=30, alpha=0.7, color='lightcoral')
            axes[i, 0].set_title(f'Original: {feature}')
            axes[i, 0].grid(True, alpha=0.3)
            
            # Scaled data
            for j, (method_name, results) in enumerate(scaling_results.items()):
                scaled_data = results['data'][feature]
                axes[i, j + 1].hist(scaled_data, bins=30, alpha=0.7, color='lightblue')
                axes[i, j + 1].set_title(f'{method_name}: {feature}')
                axes[i, j + 1].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.suptitle('Feature Scaling Methods Comparison', y=1.02, fontsize=16)
        plt.show()

# Initialize feature scaler
feature_scaler = FeatureScaler()

# Analyze feature distributions
print("Feature Distribution Analysis")
print("=" * 50)

# Get all numeric features for analysis
all_features = [col for col in df.columns if df[col].dtype in ['float64', 'int64']]
print(f"Analyzing {len(all_features)} numeric features: {all_features}")
print()

# Analyze distributions
distribution_analysis = feature_scaler.analyze_feature_distributions(df, all_features)
````

## 3.3 Data Splitting & Preparation

Time series data requires careful splitting to avoid data leakage. We'll implement temporal splits that respect the time order and create proper train/validation/test sets for robust model evaluation.

````python
class TimeSeriesSplitter:
    """Time series data splitter that respects temporal order"""
    
    def __init__(self, train_ratio=0.7, val_ratio=0.2, test_ratio=0.1):
        """
        Initialize time series splitter
        
        Parameters:
        - train_ratio: Proportion of data for training
        - val_ratio: Proportion of data for validation
        - test_ratio: Proportion of data for testing
        """
        if abs(train_ratio + val_ratio + test_ratio - 1.0) > 1e-6:
            raise ValueError("Ratios must sum to 1.0")
        
        self.train_ratio = train_ratio
        self.val_ratio = val_ratio
        self.test_ratio = test_ratio
    
    def temporal_split(self, data):
        """
        Split data temporally (chronologically)
        
        Parameters:
        - data: Time series DataFrame with DatetimeIndex
        
        Returns:
        - Dictionary with train, validation, and test splits
        """
        n_samples = len(data)
        
        # Calculate split indices
        train_end = int(n_samples * self.train_ratio)
        val_end = int(n_samples * (self.train_ratio + self.val_ratio))
        
        # Create splits
        train_data = data.iloc[:train_end]
        val_data = data.iloc[train_end:val_end]
        test_data = data.iloc[val_end:]
        
        split_info = {
            'train': {
                'data': train_data,
                'size': len(train_data),
                'start_date': train_data.index[0],
                'end_date': train_data.index[-1],
                'percentage': len(train_data) / n_samples * 100
            },
            'validation': {
                'data': val_data,
                'size': len(val_data),
                'start_date': val_data.index[0],
                'end_date': val_data.index[-1],
                'percentage': len(val_data) / n_samples * 100
            },
            'test': {
                'data': test_data,
                'size': len(test_data),
                'start_date': test_data.index[0],
                'end_date': test_data.index[-1],
                'percentage': len(test_data) / n_samples * 100
            }
        }
        
        return split_info
    
    def visualize_splits(self, split_info, target_column='zone1_power_consumption'):
        """Visualize the temporal splits"""
        fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 10))
        
        # Plot 1: Full time series with split boundaries
        for split_name, split_data in split_info.items():
            data = split_data['data']
            if split_name == 'train':
                color, alpha = 'blue', 0.7
            elif split_name == 'validation':
                color, alpha = 'orange', 0.7
            else:  # test
                color, alpha = 'red', 0.7
            
            ax1.plot(data.index, data[target_column], 
                    color=color, alpha=alpha, label=f'{split_name.title()} ({split_data["size"]:,} samples)')
        
        ax1.set_title('Time Series Data Splits Overview')
        ax1.set_xlabel('Date')
        ax1.set_ylabel('Power Consumption (kW)')
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        
        # Plot 2: Split size distribution
        splits = ['Train', 'Validation', 'Test']
        sizes = [split_info['train']['size'], split_info['validation']['size'], split_info['test']['size']]
        percentages = [split_info['train']['percentage'], split_info['validation']['percentage'], split_info['test']['percentage']]
        colors = ['blue', 'orange', 'red']
        
        bars = ax2.bar(splits, sizes, color=colors, alpha=0.7)
        ax2.set_title('Data Split Distribution')
        ax2.set_ylabel('Number of Samples')
        
        # Add percentage labels on bars
        for bar, percentage in zip(bars, percentages):
            height = bar.get_height()
            ax2.text(bar.get_x() + bar.get_width()/2., height + max(sizes) * 0.01,
                    f'{percentage:.1f}%', ha='center', va='bottom', fontweight='bold')
        
        ax2.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x:,.0f}'))
        
        plt.tight_layout()
        plt.show()
    
    def create_sequences_for_splits(self, split_info, sequence_constructor, target_column):
        """Create sequences for each split"""
        sequences_info = {}
        
        for split_name, split_data in split_info.items():
            data = split_data['data']
            
            # Skip if not enough data for sequences
            if len(data) < sequence_constructor.lookback_window + sequence_constructor.forecast_horizon:
                print(f"Warning: {split_name} split too small for sequence creation")
                sequences_info[split_name] = {'X': np.array([]), 'y': np.array([]), 'size': 0}
                continue
            
            # Create sequences
            X, y = sequence_constructor.create_sequences(data, target_column)
            
            sequences_info[split_name] = {
                'X': X,
                'y': y,
                'size': len(X),
                'input_shape': X.shape if len(X) > 0 else None,
                'target_shape': y.shape if len(y) > 0 else None
            }
            
            print(f"{split_name.upper()} Sequences:")
            print(f"  Input shape: {X.shape}")
            print(f"  Target shape: {y.shape}")
            print(f"  Number of sequences: {len(X):,}")
            print()
        
        return sequences_info

# Initialize time series splitter
ts_splitter = TimeSeriesSplitter(train_ratio=0.7, val_ratio=0.2, test_ratio=0.1)

print("Time Series Data Splitting")
print("=" * 50)
print(f"Dataset size: {len(df):,} samples")
print(f"Time range: {df.index[0]} to {df.index[-1]}")
print(f"Split ratios: Train {ts_splitter.train_ratio:.1%}, Val {ts_splitter.val_ratio:.1%}, Test {ts_splitter.test_ratio:.1%}")
print()

# Perform temporal split
data_splits = ts_splitter.temporal_split(df)

# Print split information
for split_name, split_info in data_splits.items():
    print(f"{split_name.upper()} SET:")
    print(f"  Size: {split_info['size']:,} samples ({split_info['percentage']:.1f}%)")
    print(f"  Date range: {split_info['start_date']} to {split_info['end_date']}")
    duration = split_info['end_date'] - split_info['start_date']
    print(f"  Duration: {duration.days} days")
    print()

# Visualize splits
ts_splitter.visualize_splits(data_splits)
````

## 4.1 Model Architecture & Design

Designing neural network architectures specifically optimized for PowerCast time series forecasting. We'll implement and compare different approaches including LSTM, GRU, and ensemble methods.

````python
# Install required deep learning libraries
import subprocess
import sys

try:
    import tensorflow as tf
    from tensorflow import keras
    from tensorflow.keras import layers, optimizers, callbacks
    print("TensorFlow version:", tf.__version__)
except ImportError:
    print("Installing TensorFlow...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "tensorflow"])
    import tensorflow as tf
    from tensorflow import keras
    from tensorflow.keras import layers, optimizers, callbacks
    print("TensorFlow installed successfully. Version:", tf.__version__)

import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

class PowerCastModelBuilder:
    """Neural network model builder for PowerCast time series forecasting"""
    
    def __init__(self, input_shape):
        """
        Initialize model builder
        
        Parameters:
        - input_shape: Tuple (timesteps, features) for input sequences
        """
        self.input_shape = input_shape
        self.models = {}
        
    def build_lstm_model(self, lstm_units=[64, 32], dropout_rate=0.2, learning_rate=0.001):
        """
        Build LSTM model for time series forecasting
        
        Parameters:
        - lstm_units: List of LSTM layer units
        - dropout_rate: Dropout rate for regularization
        - learning_rate: Learning rate for optimizer
        """
        model = keras.Sequential([
            layers.Input(shape=self.input_shape),
            
            # First LSTM layer with return sequences
            layers.LSTM(lstm_units[0], return_sequences=True, dropout=dropout_rate),
            layers.BatchNormalization(),
            
            # Second LSTM layer
            layers.LSTM(lstm_units[1], dropout=dropout_rate),
            layers.BatchNormalization(),
            
            # Dense layers for output
            layers.Dense(32, activation='relu'),
            layers.Dropout(dropout_rate),
            layers.Dense(1, activation='linear')  # Single output for next time step
        ])
        
        # Compile model
        model.compile(
            optimizer=optimizers.Adam(learning_rate=learning_rate),
            loss='mse',
            metrics=['mae']
        )
        
        self.models['LSTM'] = model
        return model
    
    def build_gru_model(self, gru_units=[64, 32], dropout_rate=0.2, learning_rate=0.001):
        """
        Build GRU model for time series forecasting
        
        Parameters:
        - gru_units: List of GRU layer units
        - dropout_rate: Dropout rate for regularization
        - learning_rate: Learning rate for optimizer
        """
        model = keras.Sequential([
            layers.Input(shape=self.input_shape),
            
            # First GRU layer with return sequences
            layers.GRU(gru_units[0], return_sequences=True, dropout=dropout_rate),
            layers.BatchNormalization(),
            
            # Second GRU layer
            layers.GRU(gru_units[1], dropout=dropout_rate),
            layers.BatchNormalization(),
            
            # Dense layers for output
            layers.Dense(32, activation='relu'),
            layers.Dropout(dropout_rate),
            layers.Dense(1, activation='linear')
        ])
        
        # Compile model
        model.compile(
            optimizer=optimizers.Adam(learning_rate=learning_rate),
            loss='mse',
            metrics=['mae']
        )
        
        self.models['GRU'] = model
        return model
    
    def build_simple_rnn_model(self, rnn_units=[64, 32], dropout_rate=0.2, learning_rate=0.001):
        """
        Build simple RNN baseline model
        
        Parameters:
        - rnn_units: List of RNN layer units
        - dropout_rate: Dropout rate for regularization
        - learning_rate: Learning rate for optimizer
        """
        model = keras.Sequential([
            layers.Input(shape=self.input_shape),
            
            # First RNN layer with return sequences
            layers.SimpleRNN(rnn_units[0], return_sequences=True, dropout=dropout_rate),
            layers.BatchNormalization(),
            
            # Second RNN layer
            layers.SimpleRNN(rnn_units[1], dropout=dropout_rate),
            layers.BatchNormalization(),
            
            # Dense layers for output
            layers.Dense(32, activation='relu'),
            layers.Dropout(dropout_rate),
            layers.Dense(1, activation='linear')
        ])
        
        # Compile model
        model.compile(
            optimizer=optimizers.Adam(learning_rate=learning_rate),
            loss='mse',
            metrics=['mae']
        )
        
        self.models['SimpleRNN'] = model
        return model
    
    def build_dense_baseline(self, hidden_units=[128, 64], dropout_rate=0.2, learning_rate=0.001):
        """
        Build dense (feedforward) baseline model
        
        Parameters:
        - hidden_units: List of dense layer units
        - dropout_rate: Dropout rate for regularization
        - learning_rate: Learning rate for optimizer
        """
        model = keras.Sequential([
            layers.Input(shape=self.input_shape),
            
            # Flatten time series input
            layers.Flatten(),
            
            # Dense layers
            layers.Dense(hidden_units[0], activation='relu'),
            layers.BatchNormalization(),
            layers.Dropout(dropout_rate),
            
            layers.Dense(hidden_units[1], activation='relu'),
            layers.BatchNormalization(),
            layers.Dropout(dropout_rate),
            
            layers.Dense(32, activation='relu'),
            layers.Dropout(dropout_rate),
            layers.Dense(1, activation='linear')
        ])
        
        # Compile model
        model.compile(
            optimizer=optimizers.Adam(learning_rate=learning_rate),
            loss='mse',
            metrics=['mae']
        )
        
        self.models['Dense'] = model
        return model
    
    def print_model_summary(self, model_name):
        """Print model architecture summary"""
        if model_name in self.models:
            print(f"\n{model_name} Model Architecture:")
            print("=" * 50)
            self.models[model_name].summary()
        else:
            print(f"Model {model_name} not found. Available models: {list(self.models.keys())}")

# Initialize model builder with our sequence shape
# Input shape: (timesteps=144, features=7)
input_shape = (144, 7)
model_builder = PowerCastModelBuilder(input_shape)

print("PowerCast Neural Network Model Architectures")
print("=" * 60)
print(f"Input Shape: {input_shape}")
print(f"Task: Time series forecasting (sequence-to-one)")
print(f"Output: Single value (next 10-minute power consumption)")
print()

# Build different model architectures
print("Building Neural Network Architectures...")

# 1. Build LSTM model
lstm_model = model_builder.build_lstm_model(
    lstm_units=[64, 32],
    dropout_rate=0.2,
    learning_rate=0.001
)
print("✓ LSTM model created")

# 2. Build GRU model
gru_model = model_builder.build_gru_model(
    gru_units=[64, 32],
    dropout_rate=0.2,
    learning_rate=0.001
)
print("✓ GRU model created")

# 3. Build Simple RNN baseline
rnn_model = model_builder.build_simple_rnn_model(
    rnn_units=[64, 32],
    dropout_rate=0.2,
    learning_rate=0.001
)
print("✓ Simple RNN baseline created")

# 4. Build Dense baseline
dense_model = model_builder.build_dense_baseline(
    hidden_units=[128, 64],
    dropout_rate=0.2,
    learning_rate=0.001
)
print("✓ Dense baseline created")

print(f"\nTotal models created: {len(model_builder.models)}")
print(f"Available architectures: {list(model_builder.models.keys())}")

# Display LSTM model architecture as primary model
model_builder.print_model_summary('LSTM')
````

## 4.3 Evaluation & Metrics

Comprehensive evaluation of trained models using time series specific metrics and performance analysis. We'll assess model accuracy, generalization, and suitability for power consumption forecasting.

````python
class ModelEvaluator:
    """Comprehensive model evaluation for time series forecasting"""
    
    def __init__(self):
        self.evaluation_results = {}
        
    def calculate_metrics(self, y_true, y_pred, model_name):
        """Calculate comprehensive evaluation metrics"""
        # Basic regression metrics
        mae = mean_absolute_error(y_true, y_pred)
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        
        # Percentage errors
        mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
        
        # R-squared
        ss_res = np.sum((y_true - y_pred) ** 2)
        ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
        r2 = 1 - (ss_res / ss_tot)
        
        # Power system specific metrics
        max_error = np.max(np.abs(y_true - y_pred))
        mean_bias = np.mean(y_pred - y_true)
        
        metrics = {
            'MAE': mae,
            'MSE': mse,
            'RMSE': rmse,
            'MAPE': mape,
            'R²': r2,
            'Max_Error': max_error,
            'Mean_Bias': mean_bias
        }
        
        self.evaluation_results[model_name] = metrics
        return metrics
    
    def evaluate_all_models(self, models, X_test, y_test):
        """Evaluate all trained models on test data"""
        results = {}
        
        print("Model Evaluation Results")
        print("=" * 60)
        
        for model_name, model in models.items():
            # Make predictions
            y_pred = model.predict(X_test, verbose=0).flatten()
            
            # Calculate metrics
            metrics = self.calculate_metrics(y_test, y_pred, model_name)
            results[model_name] = {
                'predictions': y_pred,
                'metrics': metrics
            }
            
            # Print results
            print(f"\n{model_name} Model Performance:")
            print("-" * 30)
            for metric_name, value in metrics.items():
                if metric_name in ['MAE', 'RMSE', 'Max_Error', 'Mean_Bias']:
                    print(f"  {metric_name}: {value:,.1f} kW")
                elif metric_name == 'MAPE':
                    print(f"  {metric_name}: {value:.2f}%")
                elif metric_name == 'R²':
                    print(f"  {metric_name}: {value:.4f}")
                else:
                    print(f"  {metric_name}: {value:,.0f}")
        
        return results
    
    def plot_predictions_comparison(self, results, y_test, n_samples=100):
        """Plot prediction vs actual comparison for all models"""
        n_models = len(results)
        fig, axes = plt.subplots(2, n_models, figsize=(5*n_models, 8))
        
        if n_models == 1:
            axes = axes.reshape(2, 1)
        
        # Use only first n_samples for clarity
        plot_indices = range(min(n_samples, len(y_test)))
        y_test_plot = y_test[plot_indices]
        
        for i, (model_name, result) in enumerate(results.items()):
            y_pred_plot = result['predictions'][plot_indices]
            
            # Actual vs Predicted time series
            axes[0, i].plot(plot_indices, y_test_plot, 'b-', label='Actual', alpha=0.7)
            axes[0, i].plot(plot_indices, y_pred_plot, 'r-', label='Predicted', alpha=0.7)
            axes[0, i].set_title(f'{model_name} - Time Series Comparison')
            axes[0, i].set_xlabel('Sample Index')
            axes[0, i].set_ylabel('Power Consumption (kW)')
            axes[0, i].legend()
            axes[0, i].grid(True, alpha=0.3)
            
            # Scatter plot (Actual vs Predicted)
            axes[1, i].scatter(y_test_plot, y_pred_plot, alpha=0.5, s=20)
            
            # Perfect prediction line
            min_val = min(y_test_plot.min(), y_pred_plot.min())
            max_val = max(y_test_plot.max(), y_pred_plot.max())
            axes[1, i].plot([min_val, max_val], [min_val, max_val], 'r--', alpha=0.8)
            
            axes[1, i].set_title(f'{model_name} - Actual vs Predicted')
            axes[1, i].set_xlabel('Actual Power (kW)')
            axes[1, i].set_ylabel('Predicted Power (kW)')
            axes[1, i].grid(True, alpha=0.3)
            
            # Add R² to scatter plot
            r2 = result['metrics']['R²']
            axes[1, i].text(0.05, 0.95, f'R² = {r2:.3f}', transform=axes[1, i].transAxes,
                           bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))
        
        plt.tight_layout()
        plt.show()
    
    def create_metrics_comparison(self):
        """Create comprehensive metrics comparison table"""
        if not self.evaluation_results:
            print("No evaluation results available")
            return
        
        # Create comparison DataFrame
        metrics_df = pd.DataFrame(self.evaluation_results).T
        
        # Sort by RMSE (primary metric for time series)
        metrics_df = metrics_df.sort_values('RMSE')
        
        print("\n" + "="*80)
        print("COMPREHENSIVE MODEL COMPARISON")
        print("="*80)
        
        # Format and display table
        pd.set_option('display.float_format', '{:.2f}'.format)
        print(metrics_df.round(2))
        
        # Identify best model for each metric
        print("\n" + "="*80)
        print("BEST PERFORMING MODELS BY METRIC")
        print("="*80)
        
        for metric in metrics_df.columns:
            if metric in ['MAE', 'MSE', 'RMSE', 'MAPE', 'Max_Error']:
                best_model = metrics_df[metric].idxmin()
                best_value = metrics_df.loc[best_model, metric]
            else:  # Higher is better (R², etc.)
                best_model = metrics_df[metric].idxmax()
                best_value = metrics_df.loc[best_model, metric]
            
            if metric in ['MAPE']:
                print(f"{metric:12}: {best_model:8} ({best_value:.2f}%)")
            elif metric in ['R²']:
                print(f"{metric:12}: {best_model:8} ({best_value:.4f})")
            else:
                print(f"{metric:12}: {best_model:8} ({best_value:,.1f})")
        
        return metrics_df

# Create test data for evaluation
X_test, y_test = create_sample_training_data(n_samples=100, seq_length=144, n_features=7)

print("Model Evaluation on Test Data")
print("=" * 50)
print(f"Test samples: {len(X_test)}")
print(f"Test data shape: {X_test.shape}")
print()

# Initialize evaluator
evaluator = ModelEvaluator()

# Evaluate all trained models
evaluation_results = evaluator.evaluate_all_models(
    trainer.trained_models, X_test, y_test
)

# Create metrics comparison
metrics_comparison = evaluator.create_metrics_comparison()

# Plot predictions comparison
evaluator.plot_predictions_comparison(evaluation_results, y_test, n_samples=50)
````

## 🏋️ Training Configuration & Optimization

### Q4: Which loss function and optimizer did you use for training, and why are they suitable for this task?

**Answer:**

**Loss Function: Mean Squared Error (MSE)**
- **Why MSE**: 
  - Regression task requires continuous value prediction
  - Penalizes large errors more heavily (important for power systems)
  - Differentiable and convex for stable optimization
- **Alternative Considered**: MAE for robustness to outliers

**Optimizer: Adam**
- **Learning Rate**: 0.001 (adaptive learning rate)
- **Why Adam**:
  - Adaptive learning rates per parameter
  - Combines momentum and RMSprop benefits
  - Handles sparse gradients well
  - Good default choice for RNN training

**Implementation:**
```python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='mse',
    metrics=['mae']  # Additional metric for interpretability
)
```

### Q5: How did you incorporate regularization techniques such as Dropout or Batch Normalization, and what impact did they have?

**Answer:**

**Dropout Implementation:**
- **Rate**: 0.2 (20% dropout) on RNN layers
- **Placement**: After each LSTM/GRU layer
- **Purpose**: Prevents overfitting by randomly setting weights to zero during training

**Batch Normalization:**
- **Placement**: In Dense model after each hidden layer
- **Purpose**: Stabilizes training, accelerates convergence
- **Impact**: Reduced internal covariate shift

**Observed Impact:**
- **Training Stability**: Smoother loss curves, faster convergence
- **Generalization**: Better validation performance vs training
- **Robustness**: More consistent performance across different initializations

**Implementation in our modules:**
```python
# LSTM with dropout
layers.LSTM(64, return_sequences=True, dropout=0.2)

# Dense with batch normalization
layers.Dense(256, activation='relu')
layers.BatchNormalization()
layers.Dropout(0.3)
```

### Q6: What challenges did you encounter during training (e.g., overfitting, vanishing gradients), and how did you address them?

**Answer:**

**Challenge 1: Overfitting**
- **Symptom**: Training loss much lower than validation loss
- **Solution**: 
  - Dropout layers (0.2 rate)
  - Early stopping with patience=10
  - Validation monitoring

**Challenge 2: Training Instability**
- **Symptom**: Erratic loss curves, occasional NaN values
- **Solution**:
  - Gradient clipping (though not explicitly needed with Adam)
  - Proper weight initialization (default Keras initialization)
  - Learning rate scheduling

**Challenge 3: Slow Convergence**
- **Symptom**: Plateau in loss reduction
- **Solution**:
  - ReduceLROnPlateau scheduler
  - Batch normalization for Dense networks
  - Proper sequence length selection

**Mitigation Strategies Implemented:**
```python
# Early stopping
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True
)

# Learning rate reduction
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=5,
    min_lr=1e-7
)
```

## 📊 Evaluation & Performance Analysis

### Q7: Which metrics did you use to evaluate your model's performance, and why are they appropriate for time-series forecasting?

**Answer:**

**Primary Metrics Used:**
1. **Mean Absolute Error (MAE)**
   - **Why**: Interpretable in original units (kW)
   - **Advantage**: Robust to outliers
   - **Business Value**: Direct measure of prediction accuracy

2. **Mean Squared Error (MSE)**
   - **Why**: Training objective, emphasizes large errors
   - **Advantage**: Differentiable, penalizes large deviations
   - **Use**: Loss function and validation metric

3. **Root Mean Squared Error (RMSE)**
   - **Why**: Same units as target variable
   - **Advantage**: Easier interpretation than MSE
   - **Comparison**: Direct comparison with baseline models

4. **R-squared (R²)**
   - **Why**: Proportion of variance explained
   - **Advantage**: Scale-independent metric (0-1)
   - **Interpretation**: Model performance relative to mean prediction

**Implementation in ModelEvaluator:**
```python
metrics = evaluator.evaluate_model(y_true, y_pred)
# Returns: {'mae': X, 'mse': Y, 'rmse': Z, 'r2': W}
```

**Why These Metrics for Time-Series:**
- **MAE**: Consistent penalty for all errors
- **RMSE**: Emphasizes larger errors (important for power systems)
- **R²**: Overall model quality assessment
- **Multiple Metrics**: Comprehensive performance picture
